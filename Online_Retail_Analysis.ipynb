# Cleaning 

```python
# Create a copy of the original dataset so we keep the raw data unchanged

df_clean = df.copy()
```

```python
# Remove exact duplicate rows to prevent the same transaction from being counted more than once
df_clean = df_clean.drop_duplicates()
```

```python
# Check how many exact duplicate rows still exist after the cleaning step
df_clean.duplicated().sum()
```

```python
# Calculate how many rows were removed during duplicate cleaning
rows_removed = len(df) - len(df_clean)

print('the number of removed rows', rows_removed)
```

```python
# Classify transactions with positive quantity and positive price as normal sales
df_clean.loc[
    (df_clean['Quantity'] > 0)
    & (df_clean['Price'] > 0),
    "Transaction_Type"
] = "Sales"
```

```python
# Count how many transactions were classified as normal sales
df_clean['Transaction_Type'].value_counts(dropna = False)
```

```python
# Classify transactions with non-positive quantity and positive price as return/cancellation candidates
df_clean.loc[
    (df_clean['Quantity'] <= 0)
    & (df_clean['Price'] > 0),
    "Transaction_Type"
] = "Return_Cancellation"
```

```python
# Check the number of transactions in each transaction type after adding the return/cancellation classification
df_clean["Transaction_Type"].value_counts(dropna=False)
```

```python
# Classify positive-quantity transactions with zero price as zero-price transactions
df_clean.loc[
    (df['Quantity'] > 0)
    & (df['Price'] == 0),
    "Transaction_Type"
] = "Zero_Price"
```

```python
# Check the transaction type distribution after adding the zero-price classification
df_clean['Transaction_Type'].value_counts(dropna = False)
```

```python
# Classify transactions with a negative price as financial adjustments
df_clean.loc[
    df_clean["Price"] < 0,
    "Transaction_Type"
] = "Financial_Adjustment"
```

```python
# Check the final distribution of transaction types after adding the financial adjustment classification
df_clean["Transaction_Type"].value_counts(dropna=False)
```

```python
# Mark transactions that do not match any confirmed business rule as unclassified
df_clean.loc[
    df_clean['Transaction_Type'].isna(),
    "Transaction_Type"
] = 'Unclassified'
```

```python
# Check the final distribution of all transaction types
df_clean["Transaction_Type"].value_counts()
```
